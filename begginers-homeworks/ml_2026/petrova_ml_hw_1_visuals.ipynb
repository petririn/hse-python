## ЗАДАНИЕ 1: Базовые графики matplotlib

```python
# Импорт библиотек
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Настройка стиля seaborn для более приятных графиков по умолчанию
sns.set_style("whitegrid")

# Настройка отображения графиков внутри ноутбука
%matplotlib inline

# Настройка размеров шрифтов для лучшей читаемости
plt.rcParams.update({
    'font.size': 12,
    'axes.titlesize': 14,
    'axes.labelsize': 12,
    'xtick.labelsize': 10,
    'ytick.labelsize': 10,
    'legend.fontsize': 10
})
```

```python
# Создаем фигуру с 4 subplots (2x2)
fig, axes = plt.subplots(2, 2, figsize=(12, 8))  # 2 строки, 2 столбца

# Название всей фигуры
fig.suptitle('Базовые графики matplotlib', fontsize=16, y=1)

# 1. Линейный график функции y = x^2 * sin(x) на интервале [0, 10]
#    Добавьте заголовок, подписи осей, сетку, легенду

# 1. Линейный график y = x^2 * sin(x)
ax = axes[0, 0]
x = np.linspace(0, 10, 100)
y = x**2 * np.sin(x)

ax.plot(x, y, 'r-', label='sin(x)', linewidth=2)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('y = x^2 * sin(x)')
ax.legend()
ax.grid(True, alpha=0.3)

# 2. Гистограмма 1000 случайных точек из нормального распределения
#    Настройте количество bins=30, добавьте заголовок

# 2. Гистограмма нормального распределения
ax = axes[0, 1]
data = np.random.normal(0, 1, 1000)
ax.hist(data, bins=30, edgecolor='black', alpha=0.7, color='skyblue')
ax.set_title('Гистограмма нормального распределения')
ax.set_xlabel('Значение')
ax.set_ylabel('Частота')
ax.grid(True, alpha=0.3)

# 3. Scatter plot: 50 точек (x - равномерное распределение, y = 2x + шум)
#    Настройте размер точек, прозрачность, цвет

# 3. Scatter plot: y = 2x + шум
ax = axes[1, 0]
x = np.random.uniform(0, 10, 50)
y = 2 * x + np.random.normal(0, 1, 50)
ax.scatter(x, y, s=60, alpha=0.6, color='green', edgecolor='black')
ax.set_title('Scatter plot: y = 2x + шум')
ax.set_xlabel('x (равномерное распределение)')
ax.set_ylabel('y')
ax.grid(True, alpha=0.3)

# 4. Bar chart: сравнение 5 разных моделей по точности
#    accuracy = [0.85, 0.92, 0.88, 0.95, 0.90]
#    Добавьте значения над столбцами

# 4. Bar chart: точность моделей
ax = axes[1, 1]
models = ['A', 'B', 'C', 'D', 'E']
accuracy = [0.85, 0.92, 0.88, 0.95, 0.90]
bars = ax.bar(models, accuracy, color='orange', edgecolor='black', alpha=0.7)
ax.set_title('Сравнение точности моделей')
ax.set_xlabel('Модель')
ax.set_ylabel('Точность')
ax.set_ylim(0.8, 1.0)
ax.grid(True, alpha=0.3)

# Добавляем значения над столбцами
for bar, acc in zip(bars, accuracy):
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height + 0.005,
            f'{acc:.2f}', ha='center')


plt.tight_layout()
plt.show()
```

## ЗАДАНИЕ 2: Seaborn визуализации

```python
# Используйте датасет tips из seaborn:
import seaborn as sns
tips = sns.load_dataset('tips')

# Создайте 3 графика:
# 1. Box plot: total_bill по дням недели (day)
#    Разделите дополнительно по времени (time) с помощью hue
# 2. Violin plot: tip в зависимости от размера компании (size)
# 3. Pair plot: выберите 4 числовых признака, раскрасьте по полу (sex)

# Загружаем датасет
tips = sns.load_dataset('tips')

# Создаем фигуру для графиков
fig, axes = plt.subplots(1, 2, figsize=(18, 6))

# 1. Box plot: total_bill по дням с разделением по времени
sns.boxplot(ax=axes[0], data=tips, x='day', y='total_bill', hue='time', palette='Set2')
axes[0].set_title('Распределение счета по дням и времени')
axes[0].set_xlabel('День недели')
axes[0].set_ylabel('Общий счет ($)')
axes[0].legend(title='Время')

# 2. Violin plot: tip по размеру компании
sns.violinplot(ax=axes[1], data=tips, x='size', y='tip', hue='size', palette='coolwarm', inner='quartile')
axes[1].set_title('Распределение чаевых по размеру компании')
axes[1].set_xlabel('Размер компании')
axes[1].set_ylabel('Чаевые ($)')


# 3. Pair plot: 4 числовых признака, раскраска по полу
# Создаем отдельную фигуру для pair plot
plt.figure(figsize=(18, 20))
pair_plot = sns.pairplot(tips[['total_bill', 'tip', 'size', 'sex']],
                         hue='sex',
                         palette={'Male': 'blue', 'Female': 'red'},
                         diag_kind='kde',
                         plot_kws={'alpha': 0.5, 's': 15},
                         height=2.5)
pair_plot.fig.suptitle('Pair plot: взаимосвязь признаков с разделением по полу', y=1.02, fontsize=14)

plt.tight_layout()
plt.show()
```
