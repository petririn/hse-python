### задача 3

```python
import pipeline_config
data = pipeline_config.pipeline_config
data
```

```python
# Включите stemming, установив "enabled": True
data['steps']['stemming']['enabled'] = True
data
```

```python
# Добавьте "numbers" в custom_words для стоп-слов
data['steps']['stopwords']['custom_words'].append('numbers')
data
```

```python
# Получите список всех включенных шагов пайплайна
enabled_steps = [
    key for key, value in data['steps'].items()
    if value.get('enabled', False)
    ]

enabled_steps
```

```python
# Измените output_format на "vectors"
data['output_format'] = 'vectors'
print(data)
```

```python
# Создайте упрощенную конфигурацию только с включенными шагами
simple_pipeline = {
    'input_encoding': data['input_encoding'],
    'output_format': data['output_format'],
    'steps': {
        step_name: {
            key: value for key, value in data['steps'][step_name].items()
            }
        for step_name in enabled_steps
        }
    }

simple_pipeline
```

### задача 4

```python
import nlp_aggegation
nlp = nlp_aggegation.models_stats
nlp
```

```python
# Найдите модель с максимальной точностью (accuracy)
max_accuracy = max(nlp.items(), key=lambda x: x[1]['accuracy'])[0]
max_accuracy
```

```python
# Рассчитайте среднее время инференса по всем моделям
times = [value['inference_time'] for value in nlp.values()]
avg_time = sum(times) / len(times)

print(times)
print(avg_time)
```

```python
# Создайте новый словарь только с метриками accuracy и f1_score для каждой модели
metrics_only = {
    key: {
        'accuracy': value['accuracy'],
        'f1_score': value['f1_score']
    }
    for key, value in nlp.items()
}

metrics_only
```

```python
# Добавьте новую модель "albert-base" с данными: accuracy=0.87, f1_score=0.86, inference_time=55, size_mb=180
nlp['albert-base'] = {
    'accuracy': 0.87,
    'f1_score': 0.86,
    'inference_time': 55,
    'size_mb': 180
}

nlp
```

```python
# Отфильтруйте модели, размер которых меньше 500 МБ
small_models = {
    key: value
    for key, value in nlp.items()
    if value['size_mb'] < 500
}

small_models
```
